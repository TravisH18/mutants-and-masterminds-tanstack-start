```python

import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://www.d20herosrd.com/6-powers/effects/effect-descriptions/"
res = requests.get(url=url)
html = res.content
soup = BeautifulSoup(html, "html.parser")
```

```python
table = soup.find_all("table")[0]
table
```

```python
rows = table.find_all("tr")
header_row = rows.pop(0)
header_data = []

for header in header_row.find_all("th"):
    header_data.append(header.get_text().lower())

header_data
```

```python
header_data.append("description")
```

```python
table_data = []
# table_headers = {0: "SKILL", 1: "ABILITY", 2: "UNTRAINED", 3: "ACTION", 4: "DESCRIPTION"}
for row in rows:
    row_data = []
    for cell in row.find_all("td"):
        print(cell)
        row_data.append(cell.get_text())
    row_data.append("")
    table_data.append(row_data)

table_data
```

```python
df = pd.DataFrame(data=table_data, columns=header_data)
df
```

```python
df.to_csv("power_effects_data.csv", index=False)
```
